```python
import numpy as np

# Input data
X = np.array([
    [1, 1],
    [1, -1],
    [-1, 1],
    [-1, -1]
])

# Target output
Y = np.array([1, -1, -1, 1])

# Initialize weights
w = np.zeros(2)
learning_rate = 1

print("Initial Weights:", w)

# Hebb Learning
for i in range(len(X)):
    w = w + learning_rate * X[i] * Y[i]
    print(f"After Sample {i+1}: {w}")

print("\nFinal Weights:", w)
```

```python
import numpy as np

X = np.array([
    [1, 1],
    [1, -1],
    [-1, 1],
    [-1, -1]
])

Y = np.array([1, -1, -1, 1])

w = np.zeros(2)

# Training
for i in range(len(X)):
    w = w + X[i] * Y[i]

print("Final Weights:", w)

print("\nPrediction")
for x in X:
    net = np.dot(x, w)
    if net >= 0:
        output = 1
    else:
        output = -1

    print("Input:", x, "Output:", output)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)

# Original graph
y = np.sin(x)

plt.plot(x, y)
plt.title("Original Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.show()
```

```python
# Add noise
noise = np.random.normal(0, 0.3, 100)

y_noisy = y + noise


plt.plot(x, y_noisy)
plt.title("Noisy Graph")
plt.grid()
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential([
    Dense(64, activation='relu', input_shape=(1,)),
    Dense(64, activation='relu'),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)
```

```python
model.fit(
    x,
    y_noisy,
    epochs=100,
    verbose=0
)
```

```python
prediction = model.predict(x)

plt.figure(figsize=(8,5))

plt.scatter(x, y_noisy, label="Noisy Data")

plt.plot(
    x,
    prediction,
    label="AI Prediction"
)

plt.plot(
    x,
    y,
    label="Original"
)

plt.title("AI Manipulated Graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid()

plt.show()
```
